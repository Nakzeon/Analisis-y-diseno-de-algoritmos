# Algoritmos de busqueda (Python)

## Busqueda lineal

Caso 1: 10 elementos

```python
import random
import time
import tracemalloc

#Uso de Gemini para esta función, prompt: Hola, como genero un arreglo con n elementos con numeros aleatorios que vayan del 1 al n?
def generar_arreglo(n):
    # Genera una lista de n números aleatorios entre 1 y n
    return [random.randint(1, n) for _ in range(n)]
def busqueda_lineal(k, arreglo):
    for i in arreglo:
        if i == k:
            return True
    return False
    
n = 10
arreglo = generar_arreglo(n)
print(arreglo)
k = -1

#Uso de Gemini para medir el consumo de memoria RAM, prompt: Y ahora, como puedo calcular el consumo de memoria de X cosa?
tracemalloc.start()
inicio = time.perf_counter()

encontrar = busqueda_lineal(k, arreglo)

fin = time.perf_counter()
memoria_actual, memoria_pico = tracemalloc.get_traced_memory()
tracemalloc.stop

total = fin - inicio


#Uso de Gemini para dar formato a los textos, prompt: Podrías darle un poco de diseño a mis textos finales?
print(f"--- Resultados para n = {n} ---")
print(f"¿Encontrado?: {encontrar}")
print(f"⏱️ Tiempo de ejecución: {total:.6f} segundos")
print(f"🧠 Memoria actual: {memoria_actual / 1024 / 1024:.2f} MB")
print(f"🚀 Pico máximo de RAM: {memoria_pico / 1024 / 1024:.2f} MB")
```

## Busqueda Centinela

Caso 1: 10 elementos

```python
import random
import time
import tracemalloc

#Uso de Gemini para esta función, prompt: Hola, como genero un arreglo con n elementos con numeros aleatorios que vayan del 1 al n?
def generar_arreglo(n):
    # Genera una lista de n números aleatorios entre 1 y n
    return [random.randint(1, n) for _ in range(n)]
def busqueda_centinela(k, arreglo):
    n = len(arreglo)
    
    # 1. Guardar el último elemento para no perderlo
    ultimo = arreglo[n - 1]
    
    # 2. Poner el centinela (el número que buscamos) al final
    arreglo[n - 1] = k
    
    i = 0
    # 3. Solo una condición: ¿es el elemento? 
    # Ya no preguntamos si i < n en cada vuelta.
    while arreglo[i] != k:
        i += 1
    
    # 4. Restauramos el último elemento original
    arreglo[n - 1] = ultimo
    
    # 5. Si i es menor al último índice, o si el último era realmente k
    if (i < n - 1) or (arreglo[n - 1] == k):
        return True
    return False
    
n = 10
arreglo = generar_arreglo(n)
print(arreglo)
k = -1

#Uso de Gemini para medir el consumo de memoria RAM, prompt: Y ahora, como puedo calcular el consumo de memoria de X cosa?
tracemalloc.start()
inicio = time.perf_counter()

encontrar = busqueda_centinela(k, arreglo)

fin = time.perf_counter()
memoria_actual, memoria_pico = tracemalloc.get_traced_memory()
tracemalloc.stop

total = fin - inicio


#Uso de Gemini para dar formato a los textos, prompt: Podrías darle un poco de diseño a mis textos finales?
print(f"--- Resultados para n = {n} ---")
print(f"¿Encontrado?: {encontrar}")
print(f"⏱️ Tiempo de ejecución: {total:.6f} segundos")
print(f"🧠 Memoria actual: {memoria_actual / 1024 / 1024:.2f} MB")
print(f"🚀 Pico máximo de RAM: {memoria_pico / 1024 / 1024:.2f} MB")
```
