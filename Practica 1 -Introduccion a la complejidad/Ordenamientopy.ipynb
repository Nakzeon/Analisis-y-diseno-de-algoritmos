# Algoritmos de ordenamiento (Python)

## Bubble Sort

Caso 1: 10 elementos

```python
import random
import time
import tracemalloc

def generar_arreglo(n):
    return [random.randint(1, n) for _ in range(n)]

def bubble_sort(arreglo):
    n = len(arreglo)
    # El primer bucle recorre todo el arreglo
    for i in range(n):
        # El segundo bucle va hasta n-i-1 porque los últimos i ya están ordenados
        for j in range(0, n - i - 1):
            # Si el elemento actual es mayor al siguiente, los intercambiamos
            if arreglo[j] > arreglo[j + 1]:
                arreglo[j], arreglo[j + 1] = arreglo[j + 1], arreglo[j]
    return arreglo

n = 10
arreglo = generar_arreglo(n)
print(arreglo)

tracemalloc.start()
inicio = time.perf_counter()

arreglo_ordenado = bubble_sort(arreglo.copy()) # .copy() para no alterar el original

fin = time.perf_counter()
memoria_actual, memoria_pico = tracemalloc.get_traced_memory()
tracemalloc.stop()

total = fin - inicio

print(f"\n--- Resultados Bubble Sort (n = {n}) ---")
print(f"⏱️ Tiempo de ejecución: {total:.6f} segundos")
print(f"🧠 Memoria actual: {memoria_actual / 1024 / 1024:.2f} MB")
print(f"🚀 Pico máximo de RAM: {memoria_pico / 1024 / 1024:.2f} MB")
```

## Merge Sort

Caso 1: 10 elementos

```python
import random
import time
import tracemalloc

def generar_arreglo(n):
    return [random.randint(1, n) for _ in range(n)]

def merge_sort(arreglo):
    if len(arreglo) > 1:
        mitad = len(arreglo) // 2
        izquierda = arreglo[:mitad]
        derecha = arreglo[mitad:]

        # Llamada recursiva para cada mitad
        merge_sort(izquierda)
        merge_sort(derecha)

        i = j = k = 0

        # Mezclamos las mitades ordenadas
        while i < len(izquierda) and j < len(derecha):
            if izquierda[i] < derecha[j]:
                arreglo[k] = izquierda[i]
                i += 1
            else:
                arreglo[k] = derecha[j]
                j += 1
            k += 1

        # Verificamos si quedaron elementos
        while i < len(izquierda):
            arreglo[k] = izquierda[i]
            i += 1
            k += 1

        while j < len(derecha):
            arreglo[k] = derecha[j]
            j += 1
            k += 1
    return arreglo

n = 10
arreglo = generar_arreglo(n)

tracemalloc.start()
inicio = time.perf_counter()

resultado = merge_sort(arreglo.copy())# .copy() para no alterar el original

fin = time.perf_counter()
memoria_actual, memoria_pico = tracemalloc.get_traced_memory()
tracemalloc.stop()

total = fin - inicio

print(f"\n--- Resultados Bubble Sort (n = {n}) ---")
print(f"⏱️ Tiempo de ejecución: {total:.6f} segundos")
print(f"🧠 Memoria actual: {memoria_actual / 1024 / 1024:.2f} MB")
print(f"🚀 Pico máximo de RAM: {memoria_pico / 1024 / 1024:.2f} MB")
```

## Bogo Sort

Caso 1: 10 elementos

```python
import random
import time
import tracemalloc

def generar_arreglo(n):
    return [random.randint(1, n) for _ in range(n)]

def esta_ordenado(arreglo):
    # Revisa si cada elemento es menor o igual al siguiente
    for i in range(len(arreglo) - 1):
        if arreglo[i] > arreglo[i + 1]:
            return False
    return True

def bogo_sort(arreglo):
    intentos = 0
    # Mientras no esté ordenado, baraja de nuevo
    while not esta_ordenado(arreglo):
        random.shuffle(arreglo)
        intentos += 1
    return arreglo, intentos

n = 8
arreglo = generar_arreglo(n)
print(arreglo)

tracemalloc.start()
inicio = time.perf_counter()

resultado, total_intentos = bogo_sort(arreglo.copy())

fin = time.perf_counter()
memoria_actual, memoria_pico = tracemalloc.get_traced_memory()
tracemalloc.stop()

total = fin - inicio

print(f"\n--- Resultados BogoSort (n = {n}) ---")
print(f"¡Ordenado en {total_intentos} intentos!")
print(f"⏱️ Tiempo de ejecución: {total:.6f} segundos")
print(f"🧠 Memoria actual: {memoria_actual / 1024 / 1024:.2f} MB")
print(f"🚀 Pico máximo de RAM: {memoria_pico / 1024 / 1024:.2f} MB")
```
