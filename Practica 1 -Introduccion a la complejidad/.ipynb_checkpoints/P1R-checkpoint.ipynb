# Algoritmos de busqueda (R)

## Busqueda lineal

Caso 1: 10 elementos

```r
# Genera una lista de n números aleatorios entre 1 y n
generar_arreglo <- function(n) {
  return(sample(1:n, n, replace = TRUE))
}

busqueda_lineal <- function(k, arreglo) {
  for (i in arreglo) {
    if (i == k) {
      return(TRUE)
    }
  }
  return(FALSE)
}

n <- 10
arreglo <- generar_arreglo(n)
print(arreglo)
k <- -1

#Uso de Gemini para obtener el consumo de RAM, prompt: Podrías tomar el método que úsaste para medir el consumo de RAM e implementarlo en R?
# --- Medición de Memoria y Tiempo ---
# gc() limpia la memoria y nos da el estado actual
# 'mem_antes' guarda la memoria usada antes de la búsqueda
mem_antes <- gc()[2, 4] # Columna de memoria usada en MB

inicio <- Sys.time()

encontrar <- busqueda_lineal(k, arreglo)

fin <- Sys.time()
mem_despues <- gc()[2, 4]

total <- as.numeric(fin - inicio)
memoria_pico <- mem_despues - mem_antes

cat(paste0("--- Resultados para n = ", n, " ---\n"))
cat(paste0("¿Encontrado?: ", encontrar, "\n"))
cat(sprintf("Tiempo de ejecución: %.6f segundos\n", total))
cat(sprintf("Memoria utilizada durante el proceso: %.2f MB\n", memoria_pico))
```

## Busqueda Centinela

Caso 1: 10 elementos

```r
#Genera una lista de n números aleatorios entre 1 y n
generar_arreglo <- function(n) {
  return(sample(1:n, n, replace = TRUE))
}

busqueda_centinela <- function(k, arreglo) {
  n <- length(arreglo)
  ultimo_original <- arreglo[n]
  arreglo[n] <- k
  i <- 1
    
  while (arreglo[i] != k) {
    i <- i + 1
  }
    
  arreglo[n] <- ultimo_original
  
  if (i < n || ultimo_original == k) {
    return(TRUE)
  }
  return(FALSE)
}

n <- 10
arreglo <- generar_arreglo(n)
print(arreglo)
k <- -1

#Uso de Gemini para obtener el consumo de RAM, prompt: Podrías tomar el método que úsaste para medir el consumo de RAM e implementarlo en R?
# --- Medición de Memoria y Tiempo ---
# gc() limpia la memoria y nos da el estado actual
# 'mem_antes' guarda la memoria usada antes de la búsqueda
mem_antes <- gc()[2, 4] 

inicio <- Sys.time()

encontrar <- busqueda_centinela(k, arreglo)

fin <- Sys.time()
mem_despues <- gc()[2, 4]

total <- as.numeric(fin - inicio)
memoria_neta <- mem_despues - mem_antes

cat(paste0("--- Resultados Busqueda Centinela n = ", n, " ---\n"))
cat(paste0("Encontrado: ", encontrar, "\n"))
cat(sprintf("Tiempo de ejecucion: %.6f segundos\n", total))
cat(sprintf("Memoria utilizada (neta): %.2f MB\n", memoria_neta))
```

```r
#Genera una lista de n números aleatorios entre 1 y n
generar_arreglo <- function(n) {
  return(sample(1:n, n, replace = TRUE))
}

busqueda_lineal_doble <- function(k, arreglo) {
  izquierda <- 1
  derecha <- length(arreglo)
  
  # El ciclo se detiene cuando los punteros se cruzan en el medio
  while (izquierda <= derecha) {
    if (arreglo[izquierda] == k) {
      return(TRUE)
    }
    if (arreglo[derecha] == k) {
      return(TRUE)
    }
    
    # Movemos los punteros hacia el centro
    izquierda <- izquierda + 1
    derecha <- derecha - 1
  }
  
  return(FALSE)
}
n <- 10
arreglo <- generar_arreglo(n)
print(arreglo)
k <- -1

#Uso de Gemini para obtener el consumo de RAM, prompt: Podrías tomar el método que úsaste para medir el consumo de RAM e implementarlo en R?
# --- Medición de Memoria y Tiempo ---
# gc() limpia la memoria y nos da el estado actual
# 'mem_antes' guarda la memoria usada antes de la búsqueda
mem_antes <- gc()[2, 4] 

inicio <- Sys.time()

encontrar <- busqueda_lineal_doble(k, arreglo)

fin <- Sys.time()
mem_despues <- gc()[2, 4]

total_tiempo <- as.numeric(fin - inicio)
memoria_neta <- mem_despues - mem_antes

# --- Resultados ---
cat(paste0("--- Resultados Busqueda Doble Extremo (n = ", n, ") ---\n"))
cat(paste0("Encontrado: ", encontrar, "\n"))
cat(sprintf("Tiempo de ejecucion: %.6f segundos\n", total_tiempo))
cat(sprintf("Memoria utilizada (neta): %.2f MB\n", memoria_neta))
```
