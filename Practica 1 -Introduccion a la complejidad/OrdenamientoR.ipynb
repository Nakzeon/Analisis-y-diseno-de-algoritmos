## Bubble Sort

Caso 1: 10 elementos

```r
generar_arreglo <- function(n) {
  return(sample(1:n, n, replace = TRUE))
}

bubble_sort <- function(arreglo) {
  n <- length(arreglo)
  
  for (i in 1:(n - 1)) {
    for (j in 1:(n - i)) {
      if (arreglo[j] > arreglo[j + 1]) {
        temp <- arreglo[j]
        arreglo[j] <- arreglo[j + 1]
        arreglo[j + 1] <- temp
      }
    }
  }
  return(arreglo)
}

n <- 10
arreglo <- generar_arreglo(n)
mem_antes <- gc()[2, 4] 

inicio <- Sys.time()
arreglo_ordenado <- bubble_sort(arreglo)

fin <- Sys.time()
mem_despues <- gc()[2, 4]

total <- as.numeric(fin - inicio)
memoria_pico <- mem_despues - mem_antes

cat(paste0("--- Resultados para n = ", n, " ---\n"))
cat(sprintf("Tiempo de ejecución: %.6f segundos\n", total))
cat(sprintf("Memoria utilizada durante el proceso: %.2f MB\n", memoria_pico))
```

## Merge Sort

Caso 1: 10 elementos

```r
#Uso de Gemini, prompt: Podrías traducir este codigo a R por favor? 
generar_arreglo <- function(n) {
  return(sample(1:n, n, replace = TRUE))
}

# Función auxiliar para mezclar las mitades
merge_parts <- function(izquierda, derecha) {
  result <- c()
  i <- 1
  j <- 1
  
  while (i <= length(izquierda) && j <= length(derecha)) {
    if (izquierda[i] <= derecha[j]) {
      result <- c(result, izquierda[i])
      i <- i + 1
    } else {
      result <- c(result, derecha[j])
      j <- j + 1
    }
  }
  
  # Agregar lo que sobre de cada lado
  if (i <= length(izquierda)) result <- c(result, izquierda[i:length(izquierda)])
  if (j <= length(derecha)) result <- c(result, derecha[j:length(derecha)])
  
  return(result)
}

merge_sort <- function(arreglo) {
  n <- length(arreglo)
  if (n <= 1) return(arreglo)
  
  medio <- floor(n / 2)
  izquierda <- merge_sort(arreglo[1:medio])
  derecha <- merge_sort(arreglo[(medio + 1):n])
  
  return(merge_parts(izquierda, derecha))
}

n <- 10
arreglo <- generar_arreglo(n)

mem_antes <- gc()[2, 4] 

inicio <- Sys.time()
arreglo_ordenado <- merge_sort(arreglo)

fin <- Sys.time()
mem_despues <- gc()[2, 4]

total <- as.numeric(fin - inicio)
memoria_pico <- mem_despues - mem_antes

cat(paste0("--- Resultados para n = ", n, " ---\n"))
cat(sprintf("Tiempo de ejecución: %.6f segundos\n", total))
cat(sprintf("Memoria utilizada durante el proceso: %.2f MB\n", memoria_pico))
```

## Bogo Sort

Caso 1: 10 elementos

```r
#Uso de Gemini, prompt: Podrías traducir este codigo a R por favor? 
generar_arreglo <- function(n) {
  return(sample(1:n, n, replace = TRUE))
}

esta_ordenado <- function(arreglo) {
  # Verifica si todos los elementos están en orden ascendente
  return(all(diff(arreglo) >= 0))
}

random_sort <- function(arreglo) {
  intentos <- 0
  # Mientras el milagro no ocurra...
  while (!esta_ordenado(arreglo)) {
    arreglo <- sample(arreglo) # Barajamos al azar
    intentos <- intentos + 1
  }
  return(list(arreglo = arreglo, intentos = intentos))
}

# --- EJECUCIÓN ---
n <- 10
arreglo <- generar_arreglo(n)

mem_antes <- gc()[2, 4] 

inicio <- Sys.time()
resultado_lista <- random_sort(arreglo)
fin <- Sys.time()

mem_despues <- gc()[2, 4]
total <- as.numeric(fin - inicio)
memoria_pico <- mem_despues - mem_antes

# --- REPORTE ---
cat(paste0("--- Resultados Random Sort (n = ", n, ") ---\n"))
cat(paste0("Intentos realizados: ", resultado_lista$intentos, "\n"))
cat(sprintf("Tiempo de ejecución: %.6f segundos\n", total))
cat(sprintf("Memoria utilizada: %.2f MB\n", memoria_pico))
```
