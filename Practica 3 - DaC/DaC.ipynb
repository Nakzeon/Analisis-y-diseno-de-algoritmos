## Divide and Conquer

Dado un arreglo A[0..n-1] de enteros (positivos y negativos), realiza lo siguiente: 

Elabora un algoritmo y un programa que calcule la suma máxima de cualquier subarreglo contiguo no vacío, utilizando el paradigma Divide y Vencerás.
Determina las complejidad temporal y espacial de tu propuesta. 
Restricciones del problema: 

1 ≤ n ≤ 10⁵
−10⁴ ≤ A[i] ≤ 10⁴
La función debe ser recursiva — sin ciclos en la parte principal
Debes implementar una función auxiliar max_cruce(A, izq, mid, der) que calcule la suma máxima de un subarreglo que cruza el punto medio

Prueba: considera los siguientes casos básicos de prueba: 

[Caso1] A=[-2, 1, -3, 4, -1, 2, 1, -5, 4] → 6 (esperado 6)<br>
[Caso2] A=[-1, -2, -3, -4] → -1 (esperado -1)<br>
[Caso3] A=[5, 4, -1, 7, 8] → 23 (esperado 23)<br>
[Caso4] A=[1] → 1 (esperado 1)<br>
[Caso5] A=[2, -1, 2, 3, 4, -1] → 10 (esperado 10)<br>

```r
max_cruce <- function(A, izq, mid, der) {
  suma_izq <- -Inf
  actual_izq <- 0
  for (i in mid:izq) {
    actual_izq <- actual_izq + A[i]
    if (actual_izq > suma_izq) {
      suma_izq <- actual_izq
    }
  }
  
  suma_der <- -Inf
  actual_der <- 0
  for (j in (mid + 1):der) {
    actual_der <- actual_der + A[j]
    if (actual_der > suma_der) {
      suma_der <- actual_der
    }
  }
  
  return(suma_izq + suma_der)
}

max_subarreglo_rec <- function(A, izq, der) {
  if (izq == der) {
    return(A[izq])
  }
  
  mid <- floor((izq + der) / 2)
  
  max_izq   <- max_subarreglo_rec(A, izq, mid)
  max_der   <- max_subarreglo_rec(A, mid + 1, der)
  max_medio <- max_cruce(A, izq, mid, der)
  
  return(max(max_izq, max_der, max_medio))
}

ejecutar_pruebas <- function() {
  casos <- list(
    list(A = c(-2, 1, -3, 4, -1, 2, 1, -5, 4), exp = 6),
    list(A = c(-1, -2, -3, -4), exp = -1),
    list(A = c(5, 4, -1, 7, 8), exp = 23),
    list(A = c(1), exp = 1),
    list(A = c(2, -1, 2, 3, -1), exp = 10)
  )
  
  for (i in seq_along(casos)) {
    res <- max_subarreglo_rec(casos[[i]]$A, 1, length(casos[[i]]$A))
    cat(sprintf("[Caso %d] Resultado: %d | Esperado: %d | %s\n", 
                i, res, casos[[i]]$exp, ifelse(res == casos[[i]]$exp, "PASÓ", "FALLÓ")))
  }
}

ejecutar_pruebas()
```
